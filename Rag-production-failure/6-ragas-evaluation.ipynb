```python
#  Imports
import os
from ragas import evaluate
from datasets import Dataset
from langchain_google_genai import ChatGoogleGenerativeAI, GoogleGenerativeAIEmbeddings
from ragas.llms import LangchainLLMWrapper
from ragas.embeddings import LangchainEmbeddingsWrapper
```

```python
# API Key
from google.colab import userdata
os.environ["GOOGLE_API_KEY"] = ("YOUR_GOOGLE_API_KEY")
```

```python
# Configure Gemini
wrapped_llm = LangchainLLMWrapper(ChatGoogleGenerativeAI(model="gemini-1.5-flash"))
wrapped_embeddings = LangchainEmbeddingsWrapper(GoogleGenerativeAIEmbeddings(model="models/embedding-001"))
```

```python
# Metrics
from ragas.metrics import faithfulness, answer_relevancy, context_precision, context_recall
metrics = [faithfulness, answer_relevancy, context_precision, context_recall]
for m in metrics:
    m.llm = wrapped_llm
    if hasattr(m, "embeddings"):
        m.embeddings = wrapped_embeddings
```

```python
#  Dataset
eval_data = {
    "question": [
        "What is the refund policy for duplicate charges?",
        "How do I update my billing information?",
        "Can I cancel my subscription mid-month?",
        "What happens if my payment fails?",
        "How long does it take to process a refund?",
    ],
    "answer": [
        "Refunds for duplicate charges are processed within 3-5 business days after the issue is confirmed.",
        "You can update your billing information by navigating to Account Settings and selecting the Billing tab.",
        "Yes, you can cancel your subscription at any time. Your access will continue until the end of the current billing period.",
        "If your payment fails, we will retry up to 3 times within 7 days before suspending your account.",
        "Standard refunds are processed within 3-5 business days. International refunds may take up to 10 business days.",
    ],
    "contexts": [
        ["Duplicate billing issue identified — customer charged twice. Refunds for duplicate charges are processed within 3-5 business days after verification by the billing team."],
        ["Go to Account > Settings > Billing to update your payment information. Changes take effect on the next billing cycle."],
        ["Subscriptions can be cancelled at any time through the Account Settings page. Service and access will continue through the end of the paid billing period."],
        ["Failed payments are retried up to 3 times over a 7-day period. If payment cannot be collected, the account will be suspended until the balance is resolved."],
        ["Refunds are typically processed within 3-5 business days for domestic accounts. International refunds may require up to 10 business days depending on the payment provider."],
    ],
    "ground_truth": [
        "Refunds for duplicate charges take 3-5 business days.",
        "Update billing info under Account > Settings > Billing.",
        "Cancel anytime; access continues through end of billing period.",
        "Failed payments are retried 3 times over 7 days, then the account is suspended.",
        "Refunds take 3-5 days domestically, up to 10 days internationally.",
    ]
}

dataset = Dataset.from_dict(eval_data)
```

```python
# Evaluate
results = evaluate(dataset, metrics=metrics)
print(results)
```
